```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your dataset
df = pd.read_csv("CarDekho_India_UsedCars_CLEANED.csv")

# Example: Create a barplot of average selling price per car brand
avg_price = df.groupby("brand")["selling_price"].mean().sort_values()

plt.figure(figsize=(12, 6))
plt.bar(avg_price.index, avg_price.values)

plt.title("Average Selling Price by Car Brand", fontsize=16, weight="bold")
plt.xlabel("Brand", fontsize=12)
plt.ylabel("Average Selling Price (INR)", fontsize=12)
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("CarDekho_India_UsedCars_CLEANED.csv")

x_col = "PUT_X_COLUMN_HERE"    # e.g., "name"
y_col = "PUT_Y_COLUMN_HERE"    # e.g., "selling_price"

summary = df.groupby(x_col)[y_col].mean().sort_values()

plt.figure(figsize=(12, 6))
plt.bar(summary.index, summary.values)

plt.title(f"{y_col} by {x_col}", fontsize=16, weight="bold")
plt.xlabel(x_col, fontsize=12)
plt.ylabel(f"Average {y_col}", fontsize=12)
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("CarDekho_India_UsedCars_CLEANED.csv")

# Group by Brand and calculate avg price
avg_price = df.groupby("Brand")["Price"].mean().sort_values()

# Plot
plt.figure(figsize=(14, 7))
plt.bar(avg_price.index, avg_price.values)

plt.title("Average Car Price by Brand", fontsize=18, weight="bold")
plt.xlabel("Brand", fontsize=14)
plt.ylabel("Average Price (INR)", fontsize=14)
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```
